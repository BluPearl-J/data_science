Find the top 5 most popular products based on the number of times they appear in the order_products table.

find the top 5 most popular products based on the number of times they appear in the order_products table. 

#using order_products__prior & order_products__train

```python
duckdb.sql("""
SELECT 
    products.product_name,
    COUNT(op.product_id) AS times_ordered
FROM (
    SELECT * FROM order_products__prior
    UNION ALL
    SELECT * FROM order_products__train
) AS op
JOIN products
  ON op.product_id = products.product_id
GROUP BY products.product_name
ORDER BY times_ordered DESC
LIMIT 5;
""").show()
```

tegorize orders into 'Weekend' (Days 0 and 1) vs 'Weekday' (Days 2-6) and count the total
orders for each category

```python
duckdb.sql("""
SELECT 
    CASE 
        WHEN order_dow IN (0,1) THEN 'Weekend'
        ELSE 'Weekday'
    END AS day_category,
    COUNT(order_id) AS total_orders
FROM orders
GROUP BY day_category
ORDER BY day_category;
""").show()
```

4 d the top 5 products that are most frequently added to the cart first (where 
add_to_cart_order = 1)

```python
duckdb.sql("""
SELECT 
    products.product_name,
    COUNT(op.product_id) AS first_added_count
FROM (
    SELECT * FROM order_products__prior
    UNION ALL
    SELECT * FROM order_products__train
) AS op
JOIN products
  ON op.product_id = products.product_id
WHERE op.add_to_cart_order = 1
GROUP BY products.product_name
ORDER BY first_added_count DESC
LIMIT 5;
""").show()
```
