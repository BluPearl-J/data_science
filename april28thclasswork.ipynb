```python
dataset_1=np.random.normal(100,10,200)
dataset_2=np.random.normal(80,30,200)
dataset_3=np.random.normal(90,20,200)
data_to_plot=[dataset_1,dataset_2,dataset_3]
plt.boxplot(data_to_plot)
plt.xticks([1,2,3],['GroupA','Group B','GroupC'])
plt.title("Comparison of Multiple Groups")
plt.xlabel("Category") 
plt.ylabel("Measurement")
plt.show()
```
