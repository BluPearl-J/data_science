```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

def data_cleaning_and_preprocessing(df):
    # --- STEP 1: THE DIVORCE ---
    X = df.drop("median_house_value", axis=1)       
    # CODE: Remove the "Answer" (Price) from the input data.
                                                    
    # WHY: You can't let the model see the answer during the test!
    
    y = df["median_house_value"].copy()            
    # CODE: Save just the "Answer" (Price) into its own variable.
                                                  
    # WHY: This is the "Gold Standard" we use to check if the model is right.

    # --- STEP 2: THE SORTING HAT ---
    num_cols = X.select_dtypes(include=[np.number]).columns.tolist() 
    # CODE: Grab all columns that are numbers.
                                                                  
    # WHY: Because you can't calculate a "median" for words.
    
    cat_cols = ["ocean_proximity"]                  
    # CODE: Explicitly list the columns that contain text.
                                                   
    # WHY: These need a "Translator" (Encoder) instead of a "Calculator."

    # --- STEP 3: THE BEAUTY PARLOR (PIPELINE) ---
    num_pipeline = Pipeline([                      
        # CODE: Glue two cleaning steps together in a specific order.
        ('imputer', SimpleImputer(strategy="median")), 
        # CODE: Plug the 207 "holes" with the middle value.
                                                     
        # WHY: Models crash on empty cells; the median is a safe "best guess."
        
        ('std_scaler', StandardScaler())           
        # CODE: Resize all numbers to a standard scale (usually -3 to 3).
                                                 
        # WHY: To stop a big number (like 50,000 population) from "bullying" a small number (like 3 bedrooms).
    ])

    # --- STEP 4: THE AIRPORT ROUTER (TRANSFORMER) ---
    full_pipeline = ColumnTransformer([          
        # CODE: Create a manager to handle different columns differently.
        ("num", num_pipeline, num_cols),          
        # CODE: Send numbers through the Beauty Parlor (Impute + Scale).
        ("cat", OneHotEncoder(), cat_cols)        
        # CODE: Send text to the Translator (Turn "INLAND" into 1s and 0s).
                                                  
        # WHY: Computers only speak "Math," so words must become numbers.
    ])

    # --- STEP 5: THE ACTION ---
    X_prepared = full_pipeline.fit_transform(X)   
    # CODE: Actually run the machines on the data.
                                                   
    # WHY: 'fit' learns the medians; 'transform' actually changes the data.
    
    return X_prepared, y, full_pipeline             
    # CODE: Give back the clean data, the labels, and the "Recipe" (Pipeline).
                                                    
    # WHY: You need that Recipe later to clean your Test Data the exact same way!
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

def data_cleaning_and_preprocessing(df):
    # 1. Separate Features (X) and Target (y)
    X = df.drop("median_house_value", axis=1)
    y = df["median_house_value"].copy()

    # 2. Identify Numeric and Categorical columns
    # We remove the target from the list of numeric columns
    num_cols = X.select_dtypes(include=[np.number]).columns.tolist()
    cat_cols = ["ocean_proximity"]

    # 3. Create a "Recipe" for Numbers (Fill gaps then Scale)
    num_pipeline = Pipeline([
        ('imputer', SimpleImputer(strategy="median")),
        ('std_scaler', StandardScaler())
    ])

    # 4. Combine everything into a ColumnTransformer
    # (Name, Transformer, Columns)
    full_pipeline = ColumnTransformer([
        ("num", num_pipeline, num_cols),
        ("cat", OneHotEncoder(), cat_cols)
    ])

    # 5. Execute the cleaning and return the prepared data
    X_prepared = full_pipeline.fit_transform(X)
    
    return X_prepared, y, full_pipeline

# --- HOW TO CALL IT ---
X_prepared, y_labels, cleaning_pipeline = data_cleaning_and_preprocessing(housing)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

def data_cleaning_and_preprocessing(df):
    # 1. Separate Features (X) and Target (y)
    X = df.drop("median_house_value", axis=1)
    y = df["median_house_value"].copy()

    # 2. Identify Numeric and Categorical columns
    # Isolate the number-only table
    numeric_only_df = X.select_dtypes(include=[np.number])
    # Extract the column names from that table
    num_cols = numeric_only_df.columns.tolist()
    # Manually list the category column
    cat_cols = ["ocean_proximity"]

    # 3. Create a "Recipe" for Numbers (Fill gaps then Scale)
    num_pipeline = Pipeline([
        ('imputer', SimpleImputer(strategy="median")),
        ('std_scaler', StandardScaler())
    ])

    # 4. Combine everything into a ColumnTransformer
    full_pipeline = ColumnTransformer([
        ("num", num_pipeline, num_cols),
        ("cat", OneHotEncoder(), cat_cols)
    ])

    # 5. Execute the cleaning and return the prepared data
    X_prepared = full_pipeline.fit_transform(X)

    return X_prepared, y, full_pipeline

# --- HOW TO CALL IT ---
X_prepared, y_labels, cleaning_pipeline = data_cleaning_and_preprocessing(housing)
```
